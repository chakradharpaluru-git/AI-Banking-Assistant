```python
# ==========================================================
# AI Banking Assistant
# Module 6 - Fraud Detection
# Step 4 - Data Cleaning
# ==========================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# ----------------------------------------------------------
# Load Dataset
# ----------------------------------------------------------

df = pd.read_csv("../datasets/fraud_data.csv")

# If running from project root, use:
# df = pd.read_csv("datasets/fraud_data.csv")

print("=" * 70)
print("STEP 4 - DATA CLEANING")
print("=" * 70)

# ----------------------------------------------------------
# Dataset Shape
# ----------------------------------------------------------

print("\nDataset Shape:")
print(df.shape)

# ==========================================================
# 1. Missing Values
# ==========================================================

print("\n" + "=" * 70)
print("1. MISSING VALUES")
print("=" * 70)

missing_values = df.isnull().sum()

print(missing_values)

total_missing = missing_values.sum()

print("\nTotal Missing Values:", total_missing)

if total_missing == 0:
    print("✅ No Missing Values Found")
else:
    print("⚠ Missing Values Detected")

# ==========================================================
# 2. Duplicate Rows
# ==========================================================

print("\n" + "=" * 70)
print("2. DUPLICATE ROWS")
print("=" * 70)

duplicates = df.duplicated().sum()

print("Duplicate Rows:", duplicates)

if duplicates == 0:
    print("✅ No Duplicate Rows Found")
else:
    print("⚠ Duplicate Rows Found")
    print("Removing duplicates...")
    df = df.drop_duplicates()
    print("New Shape:", df.shape)

# ==========================================================
# 3. Invalid Values
# ==========================================================

print("\n" + "=" * 70)
print("3. INVALID VALUES")
print("=" * 70)

# Check Amount

print("\nNegative Amount Values:")
negative_amount = (df["Amount"] < 0).sum()

print(negative_amount)

# Check Time

print("\nNegative Time Values:")
negative_time = (df["Time"] < 0).sum()

print(negative_time)

# Check Class

print("\nUnique Class Values:")
print(df["Class"].unique())

if set(df["Class"].unique()) == {0, 1}:
    print("✅ Class column is valid")
else:
    print("⚠ Invalid values found in Class column")

# ==========================================================
# 4. Outlier Detection
# ==========================================================

print("\n" + "=" * 70)
print("4. OUTLIER DETECTION")
print("=" * 70)

# Numerical Columns

numeric_columns = df.select_dtypes(include=["int64", "float64"]).columns

print("\nNumerical Columns:")
print(list(numeric_columns))

# Boxplots

for column in ["Time", "Amount"]:

    plt.figure(figsize=(8,4))

    plt.boxplot(df[column], vert=False)

    plt.title(f"Boxplot - {column}")

    plt.xlabel(column)

    plt.show()

# ==========================================================
# IQR Method
# ==========================================================

print("\nOutlier Count (IQR Method)")

for column in ["Time", "Amount"]:

    Q1 = df[column].quantile(0.25)

    Q3 = df[column].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR

    upper = Q3 + 1.5 * IQR

    outliers = df[
        (df[column] < lower) |
        (df[column] > upper)
    ]

    print(f"\n{column}")
    print("Lower Limit :", lower)
    print("Upper Limit :", upper)
    print("Outliers    :", len(outliers))

# ==========================================================
# Final Dataset Information
# ==========================================================

print("\n" + "=" * 70)
print("FINAL DATASET")
print("=" * 70)

print("Shape:", df.shape)

print("\nClass Distribution")
print(df["Class"].value_counts())

print("\n" + "=" * 70)
print("STEP 4 COMPLETED SUCCESSFULLY")
print("=" * 70)
```

```python
# ==========================================================
# AI Banking Assistant
# Module 6 - Fraud Detection
# Step 6 - Handle Imbalanced Data (SMOTE)
# ==========================================================

import pandas as pd

from sklearn.model_selection import train_test_split
from imblearn.over_sampling import SMOTE

# ----------------------------------------------------------
# Load Dataset
# ----------------------------------------------------------

df = pd.read_csv("../datasets/fraud_data.csv")

# If running from project root, use:
# df = pd.read_csv("datasets/fraud_data.csv")

print("=" * 70)
print("STEP 6 - HANDLE IMBALANCED DATA")
print("=" * 70)

# ==========================================================
# Features and Target
# ==========================================================

X = df.drop("Class", axis=1)

y = df["Class"]

print("\nOriginal Dataset Shape")
print("------------------------------")

print("Features :", X.shape)
print("Target   :", y.shape)

# ==========================================================
# Original Class Distribution
# ==========================================================

print("\nOriginal Class Distribution")
print("------------------------------")

print(y.value_counts())

# ==========================================================
# Train-Test Split
# ==========================================================

X_train, X_test, y_train, y_test = train_test_split(

    X,

    y,

    test_size=0.20,

    random_state=42,

    stratify=y

)

print("\nTraining Samples :", len(X_train))
print("Testing Samples  :", len(X_test))

# ==========================================================
# Apply SMOTE ONLY on Training Data
# ==========================================================

print("\nApplying SMOTE...")

smote = SMOTE(
    random_state=42
)

X_train_smote, y_train_smote = smote.fit_resample(

    X_train,

    y_train

)

# ==========================================================
# New Class Distribution
# ==========================================================

print("\nBalanced Training Dataset")
print("------------------------------")

print(y_train_smote.value_counts())

# ==========================================================
# Dataset Shapes
# ==========================================================

print("\nDataset Shapes")
print("------------------------------")

print("Original X_train :", X_train.shape)

print("Balanced X_train :", X_train_smote.shape)

print("Original y_train :", y_train.shape)

print("Balanced y_train :", y_train_smote.shape)

# ==========================================================
# Fraud Percentage After SMOTE
# ==========================================================

fraud = y_train_smote.value_counts()[1]

genuine = y_train_smote.value_counts()[0]

total = fraud + genuine

print("\nBalanced Dataset Percentage")
print("------------------------------")

print("Fraud    : {:.2f}%".format((fraud / total) * 100))

print("Genuine  : {:.2f}%".format((genuine / total) * 100))

# ==========================================================
# Save Balanced Dataset
# ==========================================================

balanced_df = pd.DataFrame(
    X_train_smote,
    columns=X.columns
)

balanced_df["Class"] = y_train_smote

balanced_df.to_csv(
    "../datasets/fraud_data_balanced.csv",
    index=False
)

# If running from project root:
# balanced_df.to_csv("datasets/fraud_data_balanced.csv", index=False)

print("\nBalanced Dataset Saved Successfully")

print("Location:")
print("../datasets/fraud_data_balanced.csv")

print("\n" + "=" * 70)
print("STEP 6 COMPLETED SUCCESSFULLY")
print("=" * 70)
```

```python
# ==========================================================
# AI Banking Assistant
# Module 6 - Fraud Detection
# Step 9 - Train Multiple Machine Learning Models
# ==========================================================

import warnings
warnings.filterwarnings("ignore")

import os
import joblib
import numpy as np
import pandas as pd

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report
)

from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier

# ----------------------------------------------------------
# XGBoost (Optional)
# ----------------------------------------------------------

xgb_available = True

try:
    from xgboost import XGBClassifier
except:
    xgb_available = False

# ==========================================================
# Load Processed Data
# ==========================================================

X_train = np.load(
    "../datasets/processed/fraud_X_train.npy",
    allow_pickle=True
)

X_test = np.load(
    "../datasets/processed/fraud_X_test.npy",
    allow_pickle=True
)

y_train = np.load(
    "../datasets/processed/fraud_y_train.npy",
    allow_pickle=True
)

y_test = np.load(
    "../datasets/processed/fraud_y_test.npy",
    allow_pickle=True
)

print("=" * 70)
print("PROCESSED DATA LOADED")
print("=" * 70)

print("Training Data :", X_train.shape)
print("Testing Data  :", X_test.shape)

# ==========================================================
# Models
# ==========================================================

models = {

    "Logistic Regression":
        LogisticRegression(
            max_iter=1000,
            class_weight="balanced",
            random_state=42
        ),

    "Decision Tree":
        DecisionTreeClassifier(
            class_weight="balanced",
            random_state=42
        ),

    "Random Forest":
        RandomForestClassifier(
            n_estimators=300,
            class_weight="balanced",
            random_state=42,
            n_jobs=-1
        )

}

# ----------------------------------------------------------
# Add XGBoost (if installed)
# ----------------------------------------------------------

if xgb_available:

    models["XGBoost"] = XGBClassifier(

        n_estimators=300,

        learning_rate=0.1,

        max_depth=6,

        random_state=42,

        eval_metric="logloss"

    )

# ==========================================================
# Train Models
# ==========================================================

results = []

best_model = None

best_model_name = ""

best_f1 = 0

print("\n")
print("=" * 70)
print("MODEL TRAINING STARTED")
print("=" * 70)

for name, model in models.items():

    print(f"\nTraining : {name}")

    model.fit(X_train, y_train)

    prediction = model.predict(X_test)

    accuracy = accuracy_score(y_test, prediction)

    precision = precision_score(
        y_test,
        prediction,
        zero_division=0
    )

    recall = recall_score(
        y_test,
        prediction,
        zero_division=0
    )

    f1 = f1_score(
        y_test,
        prediction,
        zero_division=0
    )

    results.append([
        name,
        accuracy,
        precision,
        recall,
        f1
    ])

    print(f"Accuracy : {accuracy:.4f}")
    print(f"Precision: {precision:.4f}")
    print(f"Recall   : {recall:.4f}")
    print(f"F1 Score : {f1:.4f}")

    print("\nConfusion Matrix")
    print(confusion_matrix(y_test, prediction))

    print("\nClassification Report")
    print(classification_report(y_test, prediction))

    if f1 > best_f1:

        best_f1 = f1
        best_model = model
        best_model_name = name

# ==========================================================
# Model Comparison
# ==========================================================

results_df = pd.DataFrame(

    results,

    columns=[

        "Model",

        "Accuracy",

        "Precision",

        "Recall",

        "F1 Score"

    ]

)

results_df = results_df.sort_values(
    by="F1 Score",
    ascending=False
)

print("\n")
print("=" * 70)
print("MODEL COMPARISON")
print("=" * 70)

print(results_df)

# ==========================================================
# Save Best Model
# ==========================================================

os.makedirs("../models", exist_ok=True)

joblib.dump(
    best_model,
    "../models/fraud_model.pkl"
)

print("\n")
print("=" * 70)
print("BEST MODEL SAVED")
print("=" * 70)

print("Best Model :", best_model_name)
print("Best F1 Score :", round(best_f1, 4))

print("\nSaved Location:")
print("../models/fraud_model.pkl")
```

```python
# ==========================================================
# AI Banking Assistant
# Module 6 - Fraud Detection
# Step 10 - Evaluate Models
# ==========================================================

import warnings
warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix,
    ConfusionMatrixDisplay,
    classification_report
)

import joblib

# ==========================================================
# Load Test Data
# ==========================================================

X_test = np.load(
    "../datasets/processed/fraud_X_test.npy",
    allow_pickle=True
)

y_test = np.load(
    "../datasets/processed/fraud_y_test.npy",
    allow_pickle=True
)

# ==========================================================
# Load Best Model
# ==========================================================

model = joblib.load(
    "../models/fraud_model.pkl"
)

print("=" * 70)
print("MODEL EVALUATION")
print("=" * 70)

# ==========================================================
# Predictions
# ==========================================================

y_pred = model.predict(X_test)

# Probability Scores (for ROC-AUC)

if hasattr(model, "predict_proba"):

    y_prob = model.predict_proba(X_test)[:,1]

elif hasattr(model, "decision_function"):

    scores = model.decision_function(X_test)

    y_prob = (
        scores - scores.min()
    ) / (
        scores.max() - scores.min()
    )

else:

    y_prob = y_pred

# ==========================================================
# Metrics
# ==========================================================

accuracy = accuracy_score(
    y_test,
    y_pred
)

precision = precision_score(
    y_test,
    y_pred,
    zero_division=0
)

recall = recall_score(
    y_test,
    y_pred,
    zero_division=0
)

f1 = f1_score(
    y_test,
    y_pred,
    zero_division=0
)

roc_auc = roc_auc_score(
    y_test,
    y_prob
)

# ==========================================================
# Print Metrics
# ==========================================================

print("\nEvaluation Metrics")
print("-" * 70)

print(f"Accuracy : {accuracy:.4f}")
print(f"Precision: {precision:.4f}")
print(f"Recall   : {recall:.4f}")
print(f"F1 Score : {f1:.4f}")
print(f"ROC-AUC  : {roc_auc:.4f}")

# ==========================================================
# Classification Report
# ==========================================================

print("\nClassification Report")
print("-" * 70)

print(
    classification_report(
        y_test,
        y_pred
    )
)

# ==========================================================
# Confusion Matrix
# ==========================================================

cm = confusion_matrix(
    y_test,
    y_pred
)

print("\nConfusion Matrix")
print("-" * 70)

print(cm)

# ==========================================================
# Plot Confusion Matrix
# ==========================================================

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=["Genuine", "Fraud"]
)

disp.plot(
    cmap="Blues"
)

plt.title("Confusion Matrix")

plt.show()

# ==========================================================
# Summary
# ==========================================================

print("\n" + "=" * 70)
print("MODEL PERFORMANCE SUMMARY")
print("=" * 70)

summary = pd.DataFrame({

    "Metric":[

        "Accuracy",
        "Precision",
        "Recall",
        "F1 Score",
        "ROC-AUC"

    ],

    "Value":[

        accuracy,
        precision,
        recall,
        f1,
        roc_auc

    ]

})

print(summary)

print("\n" + "=" * 70)
print("IMPORTANT OBSERVATION")
print("=" * 70)

print("""
Fraud Detection datasets are highly imbalanced.

Accuracy alone can be misleading.

Always focus on:

✔ Precision
✔ Recall
✔ F1 Score
✔ ROC-AUC

These metrics better reflect the model's ability
to detect fraudulent transactions.
""")

print("=" * 70)
print("STEP 10 COMPLETED SUCCESSFULLY")
print("=" * 70)
```
