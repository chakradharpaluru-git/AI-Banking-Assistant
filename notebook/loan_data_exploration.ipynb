```python
# ==========================================
# AI Banking Assistant
# Module 1 - Loan Dataset Exploration
# ==========================================

import os
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# ------------------------------------------
# Show Current Working Directory
# ------------------------------------------

print("Current Working Directory:")
print(os.getcwd())

# ------------------------------------------
# Dataset Path
# ------------------------------------------
# If your notebook is in the project root:
dataset_path = r"C:\Users\aswin\OneDrive\Desktop\AI-Banking-Assistant\datasets\loan_data.csv"

# If your notebook is inside a "notebooks" folder,
# comment the above line and use this instead:
# dataset_path = "../datasets/loan_data.csv"

# ------------------------------------------
# Load Dataset
# ------------------------------------------

if not os.path.exists(dataset_path):
    print(f"\nDataset not found: {dataset_path}")
    print("Please place loan_data.csv in the datasets folder.")
else:
    df = pd.read_csv(dataset_path)

    print("\nDataset Loaded Successfully!")

    # ------------------------------------------
    # First 5 Rows
    # ------------------------------------------
    print("\n========== FIRST 5 ROWS ==========")
    print(df.head())

    # ------------------------------------------
    # Shape
    # ------------------------------------------
    print("\n========== SHAPE ==========")
    print(df.shape)

    # ------------------------------------------
    # Columns
    # ------------------------------------------
    print("\n========== COLUMNS ==========")
    print(df.columns.tolist())

    # ------------------------------------------
    # Dataset Info
    # ------------------------------------------
    print("\n========== INFO ==========")
    print(df.info())

    # ------------------------------------------
    # Statistics
    # ------------------------------------------
    print("\n========== STATISTICS ==========")
    print(df.describe())

    # ------------------------------------------
    # Missing Values
    # ------------------------------------------
    print("\n========== MISSING VALUES ==========")
    print(df.isnull().sum())

    # ------------------------------------------
    # Duplicate Rows
    # ------------------------------------------
    print("\n========== DUPLICATES ==========")
    print(df.duplicated().sum())

    # ------------------------------------------
    # Loan Status
    # ------------------------------------------
    print("\n========== LOAN STATUS ==========")
    print(df["Loan_Status"].value_counts())

    # ------------------------------------------
    # Numerical Columns
    # ------------------------------------------
    print("\n========== NUMERICAL COLUMNS ==========")
    print(df.select_dtypes(include=["int64", "float64"]).columns.tolist())

    # ------------------------------------------
    # Categorical Columns
    # ------------------------------------------
    print("\n========== CATEGORICAL COLUMNS ==========")
    print(df.select_dtypes(include=["object"]).columns.tolist())

    # ------------------------------------------
    # Loan Status Plot
    # ------------------------------------------
    plt.figure(figsize=(6,4))
    sns.countplot(x="Loan_Status", data=df)
    plt.title("Loan Approval Distribution")
    plt.show()

    # ------------------------------------------
    # Gender Plot
    # ------------------------------------------
    plt.figure(figsize=(6,4))
    sns.countplot(x="Gender", data=df)
    plt.title("Gender Distribution")
    plt.show()

    # ------------------------------------------
    # Education Plot
    # ------------------------------------------
    plt.figure(figsize=(6,4))
    sns.countplot(x="Education", data=df)
    plt.title("Education Distribution")
    plt.show()

    # ------------------------------------------
    # Property Area Plot
    # ------------------------------------------
    plt.figure(figsize=(6,4))
    sns.countplot(x="Property_Area", data=df)
    plt.title("Property Area")
    plt.show()

    # ------------------------------------------
    # Applicant Income Histogram
    # ------------------------------------------
    plt.figure(figsize=(8,5))
    plt.hist(df["ApplicantIncome"], bins=30)
    plt.title("Applicant Income Distribution")
    plt.xlabel("Applicant Income")
    plt.ylabel("Count")
    plt.show()

    # ------------------------------------------
    # Loan Amount Histogram
    # ------------------------------------------
    plt.figure(figsize=(8,5))
    plt.hist(df["LoanAmount"].dropna(), bins=30)
    plt.title("Loan Amount Distribution")
    plt.xlabel("Loan Amount")
    plt.ylabel("Count")
    plt.show()

    # ------------------------------------------
    # Create datasets folder if it doesn't exist
    # ------------------------------------------
    os.makedirs("datasets", exist_ok=True)

    # ------------------------------------------
    # Save Copy
    # ------------------------------------------
    save_path = "datasets/loan_data_copy.csv"

    df.to_csv(save_path, index=False)

    print("\nDataset copy saved successfully!")
    print(f"Saved at: {os.path.abspath(save_path)}")

    print("\n======================================")
    print("DATA EXPLORATION COMPLETED SUCCESSFULLY")
    print("======================================")
```
